# 4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.
<li> Calculate the frequency of each word and save the output in CSV format (Term, Frequency) </li> 
<li> Identify the top 10 most frequent terms and display their corresponding frequencies. </li>

```python
import nltk
import pandas as pd
import spacy
from collections import Counter

# Download tokenizer
nltk.download("punkt")

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Read input from input_text.txt

with open("input_text.txt", "r", encoding="utf-8") as file:
    text = file.read()

print("Input Text:")
print(text)

# Term Frequency Analysis


# Convert to lowercase and tokenize
words = nltk.word_tokenize(text.lower())

# Keep only words
words = [word for word in words if word.isalpha()]

# Calculate frequency
frequency = Counter(words)

# Create DataFrame
df = pd.DataFrame(
    frequency.items(),
    columns=["Term", "Frequency"]
)

# Sort by frequency
df = df.sort_values(
    by="Frequency",
    ascending=False
)

# Save to CSV
df.to_csv("term_frequency.csv", index=False)

print("\nTerm-Frequency Analysis")
print(df)

# Top 10 Most Frequent Terms

print("\nTop 10 Most Frequent Terms")
print("--------------------------")

top_10 = df.head(10)

for _, row in top_10.iterrows():
    print(f"{row['Term']} : {row['Frequency']}")

# Named Entity Recognition

doc = nlp(text)

print("\nNamed Entities")
print("--------------------------")

for entity in doc.ents:
    print(f"{entity.text} -> {entity.label_}")
```

# 4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.
<li>Calculate the frequency of each word and save the output in CSV format (Term,
Frequency) </li>
<li>Identify the top 10 most frequent terms and display their corresponding
frequencies. <li>
<b>Input data link of 4.1. And 4.2. :
<link>https://drive.google.com/file/d/14DC156guTGLzGDbMGGoIUvlhBbCu_jLZ/view?usp=sharing</link></b>


```python
import re
import csv

# Read the input text file
with open("input_text.txt", "r", encoding="utf-8") as file:
    text = file.read()

print("Input Text:")
print(text)
```

```python
# Convert text to lowercase
text = text.lower()

# Extract words using regular expression
words = re.findall(r'\b[a-zA-Z]+\b', text)

# Calculate frequency manually
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Sort terms by frequency in descending order
sorted_frequency = sorted(
    frequency.items(),
    key=lambda x: x[1],
    reverse=True
)

# Display frequency of each word
print("\nTerm Frequency:")
print("----------------------------")

for term, count in sorted_frequency:
    print(f"{term} : {count}")
```

```python
# Save Term-Frequency results to CSV
with open("term_frequency.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    # Header
    writer.writerow(["Term", "Frequency"])

    # Data
    for term, count in sorted_frequency:
        writer.writerow([term, count])

print("\nTerm-frequency results saved successfully!")
print("File: term_frequency.csv")
```

```python
print("\nTop 10 Most Frequent Terms")
print("==========================")

top_10 = sorted_frequency[:10]

for i, (term, count) in enumerate(top_10, start=1):
    print(f"{i}. {term} : {count}")
```

# 4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:

1. Read multiple text documents.
2. Convert text to lowercase.
3. Remove punctuation.
4. Tokenize the documents.
5. Calculate TF for every term.
6. Calculate DF for every term.
7. Calculate IDF using the logarithmic formula.
8. Calculate TF-IDF for every term in every document.
9. Display Term, TF, DF, IDF, and TF-IDF at each step.
10. Identify the top 10 terms with the highest TF-IDF score for each document.

Given Input
Consider the following three documents:
Document 1:
Natural language processing is a field of artificial intelligence.
Document 2:
Natural language processing helps computers understand human language.
Document 3:
Machine learning is an important part of artificial intelligence.

```python
import math
import string
```

```python
# Step 1: Given Documents

documents = {
    "Document 1": "Natural language processing is a field of artificial intelligence.",
    
    "Document 2": "Natural language processing helps computers understand human language.",
    
    "Document 3": "Machine learning is an important part of artificial intelligence."
}
```

```python
# Step 2: Convert text to lowercase

lowercase_documents = {}

for doc_name, text in documents.items():
    lowercase_documents[doc_name] = text.lower()

print("=" * 70)
print("STEP 2: LOWERCASE TEXT")
print("=" * 70)

for doc_name, text in lowercase_documents.items():
    print(f"{doc_name}:")
    print(text)
    print()
```

```python
# Step 3: Remove punctuation

cleaned_documents = {}

for doc_name, text in lowercase_documents.items():

    # Replace punctuation with nothing
    text = text.translate(str.maketrans("", "", string.punctuation))

    cleaned_documents[doc_name] = text


print("=" * 70)
print("STEP 3: TEXT AFTER REMOVING PUNCTUATION")
print("=" * 70)

for doc_name, text in cleaned_documents.items():
    print(f"{doc_name}:")
    print(text)
    print()
```

```python
# Step 4: Tokenization

tokenized_documents = {}

for doc_name, text in cleaned_documents.items():

    # Simple tokenization using split()
    tokens = text.split()

    tokenized_documents[doc_name] = tokens


print("=" * 70)
print("STEP 4: TOKENIZED DOCUMENTS")
print("=" * 70)

for doc_name, tokens in tokenized_documents.items():
    print(f"{doc_name}:")
    print(tokens)
    print()
```

```python
# Step 5: Calculate TF

tf = {}

for doc_name, tokens in tokenized_documents.items():

    tf[doc_name] = {}

    total_terms = len(tokens)

    # Count occurrences of every term
    term_counts = {}

    for token in tokens:

        if token in term_counts:
            term_counts[token] += 1
        else:
            term_counts[token] = 1

    # Calculate normalized TF
    for term, count in term_counts.items():
        tf[doc_name][term] = count / total_terms


print("=" * 70)
print("STEP 5: TERM FREQUENCY (TF)")
print("=" * 70)

for doc_name in tf:

    print(f"\n{doc_name}")

    print(f"{'Term':<20}{'TF':>10}")
    print("-" * 30)

    for term, value in sorted(tf[doc_name].items()):
        print(f"{term:<20}{value:>10.4f}")
```

```python
# Step 6: Calculate DF

df = {}

for doc_name, tokens in tokenized_documents.items():

    # Use set so a term is counted only once per document
    unique_terms = set(tokens)

    for term in unique_terms:

        if term in df:
            df[term] += 1
        else:
            df[term] = 1


print("\n")
print("=" * 70)
print("STEP 6: DOCUMENT FREQUENCY (DF)")
print("=" * 70)

print(f"{'Term':<20}{'DF':>10}")
print("-" * 30)

for term in sorted(df):
    print(f"{term:<20}{df[term]:>10}")
```

```python
# ----------------------------------------------------------
# Step 7: Calculate IDF
# ----------------------------------------------------------

number_of_documents = len(documents)

idf = {}

for term, document_frequency in df.items():

    idf[term] = math.log(
        number_of_documents / document_frequency
    )


print("\n")
print("=" * 70)
print("STEP 7: INVERSE DOCUMENT FREQUENCY (IDF)")
print("=" * 70)

print(f"{'Term':<20}{'DF':>10}{'IDF':>15}")
print("-" * 45)

for term in sorted(idf):

    print(
        f"{term:<20}"
        f"{df[term]:>10}"
        f"{idf[term]:>15.4f}"
    )
```

```python
# ----------------------------------------------------------
# Step 8: Calculate TF-IDF
# ----------------------------------------------------------

tfidf = {}

for doc_name in tokenized_documents:

    tfidf[doc_name] = {}

    for term in tf[doc_name]:

        tfidf[doc_name][term] = (
            tf[doc_name][term] * idf[term]
        )
```

```python
# ----------------------------------------------------------
# Step 9: Display TF, DF, IDF and TF-IDF
# ----------------------------------------------------------

print("\n")
print("=" * 70)
print("STEP 9: TF, DF, IDF AND TF-IDF")
print("=" * 70)

for doc_name in tfidf:

    print(f"\n{doc_name}")

    print(
        f"{'Term':<20}"
        f"{'TF':>10}"
        f"{'DF':>10}"
        f"{'IDF':>15}"
        f"{'TF-IDF':>15}"
    )

    print("-" * 70)

    for term in sorted(tfidf[doc_name]):

        print(
            f"{term:<20}"
            f"{tf[doc_name][term]:>10.4f}"
            f"{df[term]:>10}"
            f"{idf[term]:>15.4f}"
            f"{tfidf[doc_name][term]:>15.4f}"
        )
```

```python

# ----------------------------------------------------------
# Step 10: Top 10 TF-IDF Terms
# ----------------------------------------------------------

print("\n")
print("=" * 70)
print("STEP 10: TOP 10 TERMS BY TF-IDF")
print("=" * 70)

for doc_name in tfidf:

    # Sort terms according to TF-IDF score
    sorted_terms = sorted(
        tfidf[doc_name].items(),
        key=lambda x: x[1],
        reverse=True
    )

    print(f"\n{doc_name}")
    print("-" * 40)

    for rank, (term, score) in enumerate(
        sorted_terms[:10], start=1
    ):

        print(
            f"{rank}. {term:<20} {score:.4f}"
        )
```
