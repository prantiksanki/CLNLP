# Part B — BPE Without Using a Pretrained Model

```python
import re
from collections import Counter
import pandas as pd
```

```python
with open("input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read().strip()

print("Input Text:")
print(text)
```

```python
def get_words(text):
    """
    Extract words and punctuation from the input text.
    """

    return re.findall(
        r"[A-Za-z0-9]+(?:['-][A-Za-z0-9]+)*|[^A-Za-z0-9\s]",
        text.lower()
    )


words = get_words(text)

print("Words:")
print(words)

print("\nNumber of words:", len(words))
```

```python
def build_initial_vocab(words):

    vocab = Counter()

    for word in words:

        # Convert word into characters
        symbols = tuple(word) + ("</w>",)

        # Count frequency
        vocab[symbols] += 1

    return vocab


vocab = build_initial_vocab(words)

print("Initial Vocabulary:\n")

for word, frequency in vocab.items():
    print(word, "->", frequency)
```

```python
def get_pair_counts(vocab):

    pairs = Counter()

    for symbols, frequency in vocab.items():

        for i in range(len(symbols) - 1):

            pair = (
                symbols[i],
                symbols[i + 1]
            )

            pairs[pair] += frequency

    return pairs


pair_counts = get_pair_counts(vocab)

print("Most frequent symbol pairs:\n")

for pair, frequency in pair_counts.most_common(20):

    print(
        f"{pair} -> {frequency}"
    )
```

```python
def merge_pair(pair, vocab):

    first, second = pair

    new_vocab = Counter()

    for symbols, frequency in vocab.items():

        new_symbols = []

        i = 0

        while i < len(symbols):

            # Check whether current and next symbols
            # are the pair that needs to be merged

            if (
                i < len(symbols) - 1
                and symbols[i] == first
                and symbols[i + 1] == second
            ):

                new_symbols.append(first + second)

                i += 2

            else:

                new_symbols.append(symbols[i])

                i += 1

        new_vocab[tuple(new_symbols)] += frequency

    return new_vocab
```

```python
def train_bpe(text, num_merges=100):

    words = get_words(text)

    vocab = build_initial_vocab(words)

    merges = []

    for iteration in range(num_merges):

        pair_counts = get_pair_counts(vocab)

        # Stop if no pairs are available
        if not pair_counts:
            break

        # Find most frequent pair
        best_pair, best_frequency = pair_counts.most_common(1)[0]

        # Stop if no pair occurs more than once
        if best_frequency < 2:
            break

        # Save merge rule
        merges.append(best_pair)

        # Perform merge
        vocab = merge_pair(
            best_pair,
            vocab
        )

    return vocab, merges
```

```python
final_vocab, merges = train_bpe(
    text,
    num_merges=100
)

print("Number of learned merge rules:", len(merges))

print("\nFirst 20 BPE merge rules:\n")

for i, pair in enumerate(merges[:20], start=1):

    first, second = pair

    print(
        f"{i}. {first!r} + {second!r}"
        f" -> {(first + second)!r}"
    )
```

```python
print("Final BPE Vocabulary:\n")

learned_tokens = set()

for symbols in final_vocab:

    for symbol in symbols:

        learned_tokens.add(symbol)


for i, token in enumerate(
    sorted(learned_tokens),
    start=1
):

    print(i, repr(token))

print(
    "\nVocabulary Size:",
    len(learned_tokens)
)
```

```python
def apply_bpe_to_word(word, merges):

    # Start with individual characters
    symbols = list(
        tuple(word.lower()) + ("</w>",)
    )

    # Apply learned merges in order
    for pair in merges:

        first, second = pair

        merged_symbols = []

        i = 0

        while i < len(symbols):

            if (
                i < len(symbols) - 1
                and symbols[i] == first
                and symbols[i + 1] == second
            ):

                merged_symbols.append(
                    first + second
                )

                i += 2

            else:

                merged_symbols.append(
                    symbols[i]
                )

                i += 1

        symbols = merged_symbols

    return symbols
```

```python
def tokenize_text(text, merges):

    tokens = []

    # Keep words, punctuation and whitespace
    parts = re.findall(
        r"[A-Za-z0-9]+(?:['-][A-Za-z0-9]+)*"
        r"|[^A-Za-z0-9\s]"
        r"|\s+",
        text
    )

    for part in parts:

        # Whitespace
        if part.isspace():

            tokens.append("<SPACE>")

        # Word
        elif re.fullmatch(
            r"[A-Za-z0-9]+(?:['-][A-Za-z0-9]+)*",
            part
        ):

            word_tokens = apply_bpe_to_word(
                part,
                merges
            )

            tokens.extend(word_tokens)

        # Punctuation
        else:

            tokens.append(part)

    return tokens
```

```python
tokens_from_scratch = tokenize_text(
    text,
    merges
)

print("Generated BPE Tokens:\n")

for i, token in enumerate(
    tokens_from_scratch,
    start=1
):

    print(
        f"{i:4d}. {token!r}"
    )
```

```python
def create_token_ids(tokens):

    # Get unique tokens
    unique_tokens = sorted(
        set(tokens)
    )

    # Reserve ID 0 for unknown token
    token_to_id = {
        "<UNK>": 0
    }

    # Assign IDs
    for token in unique_tokens:

        if token not in token_to_id:

            token_to_id[token] = len(
                token_to_id
            )

    # Convert tokens to IDs
    token_ids = [
        token_to_id.get(token, 0)
        for token in tokens
    ]

    return token_to_id, token_ids


token_to_id, token_ids_from_scratch = create_token_ids(
    tokens_from_scratch
)
```

```python
print("=" * 60)
print("BPE FROM SCRATCH")
print("=" * 60)

print(
    f"{'TOKEN':<25} {'TOKEN ID'}"
)

print("-" * 60)

for token, token_id in zip(
    tokens_from_scratch,
    token_ids_from_scratch
):

    print(
        f"{token!r:<25} {token_id}"
    )

print("=" * 60)

print(
    "Total Tokens:",
    len(tokens_from_scratch)
)

print(
    "Vocabulary Size:",
    len(token_to_id)
)
```

```python
scratch_table = pd.DataFrame({
    "Token": tokens_from_scratch,
    "Token ID": token_ids_from_scratch
})

scratch_table
```

```python
print("=" * 70)
print("COMPARISON OF TWO BPE APPROACHES")
print("=" * 70)

print("\n1. PRETRAINED GPT-2 BPE")
print("-------------------------")
print("Number of tokens:", len(tokens))
print("Vocabulary is pretrained: Yes")

print("\n2. BPE FROM SCRATCH")
print("-------------------------")
print("Number of tokens:", len(tokens_from_scratch))
print("Vocabulary is pretrained: No")
print("Learned merge rules:", len(merges))
print("Vocabulary size:", len(token_to_id))
```

```python
# Tokenize the input text

tokens = tokenizer.tokenize(text)

print("Generated SentencePiece Tokens:\n")

for i, token in enumerate(tokens, start=1):
    print(f"{i:4d}. {token!r}")
```

```python
# Convert tokens into their corresponding token IDs

token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Token IDs:\n")

for token, token_id in zip(tokens, token_ids):
    print(f"{token!r:<25} -> {token_id}")
```

```python
# Decode the token IDs back into text

decoded_text = tokenizer.decode(
    encoded_ids,
    skip_special_tokens=True
)

print("Original Text:")
print(text)

print("\nDecoded Text:")
print(decoded_text)
```

# Part A — POS Tagging Using spaCy

```python
!pip install spacy -q
```

```python
import spacy
import pandas as pd
```

```python
# Load the pretrained English language model

nlp = spacy.load("en_core_web_sm")

print("spaCy English model loaded successfully.")
```

```python
text = "The young student is reading an interesting book in the library."

print("Input Text:")
print(text)
```

```python
# Process the sentence using spaCy

doc = nlp(text)

print("POS Tagged Tokens:\n")

for token in doc:
    print(
        token.text,
        "->",
        token.pos_,
        "->",
        spacy.explain(token.pos_)
    )
```

```python
# Dictionary to store unique tokens
unique_tokens = {}

for token in doc:

    # Ignore punctuation
    if token.is_punct:
        continue

    token_text = token.text

    # Store only the first occurrence
    if token_text not in unique_tokens:

        unique_tokens[token_text] = {
            "POS Tag": token.pos_,
            "Description": spacy.explain(token.pos_)
        }


# Convert dictionary into DataFrame

spacy_table = pd.DataFrame([
    {
        "Token": token,
        "POS Tag": information["POS Tag"],
        "Description": information["Description"]
    }
    for token, information in unique_tokens.items()
])

spacy_table
```

```python
print("=" * 75)
print("POS TAGGING USING SPACY")
print("=" * 75)

print(
    f"{'TOKEN':<20}"
    f"{'POS TAG':<15}"
    f"{'DESCRIPTION'}"
)

print("-" * 75)

for _, row in spacy_table.iterrows():

    print(
        f"{row['Token']:<20}"
        f"{row['POS Tag']:<15}"
        f"{row['Description']}"
    )

print("-" * 75)

print("Number of unique tokens:", len(spacy_table))
```

```python
# Tokenize the sentence into individual words

tokens = word_tokenize(text)

print("Tokens:")
print(tokens)
```

```python
# Apply NLTK POS tagging

tagged_tokens = pos_tag(tokens)

print("POS Tagged Tokens:\n")

for token, tag in tagged_tokens:
    print(
        token,
        "->",
        tag
    )
```

```python
pos_descriptions = {
    "CC": "Coordinating conjunction",
    "CD": "Cardinal number",
    "DT": "Determiner",
    "EX": "Existential there",
    "FW": "Foreign word",
    "IN": "Preposition or subordinating conjunction",
    "JJ": "Adjective",
    "JJR": "Adjective, comparative",
    "JJS": "Adjective, superlative",
    "LS": "List item marker",
    "MD": "Modal",
    "NN": "Noun, singular or mass",
    "NNS": "Noun, plural",
    "NNP": "Proper noun, singular",
    "NNPS": "Proper noun, plural",
    "PDT": "Predeterminer",
    "POS": "Possessive ending",
    "PRP": "Personal pronoun",
    "PRP$": "Possessive pronoun",
    "RB": "Adverb",
    "RBR": "Adverb, comparative",
    "RBS": "Adverb, superlative",
    "RP": "Particle",
    "SYM": "Symbol",
    "TO": "To",
    "UH": "Interjection",
    "VB": "Verb, base form",
    "VBD": "Verb, past tense",
    "VBG": "Verb, gerund or present participle",
    "VBN": "Verb, past participle",
    "VBP": "Verb, non-3rd person singular present",
    "VBZ": "Verb, 3rd person singular present",
    "WDT": "Wh-determiner",
    "WP": "Wh-pronoun",
    "WP$": "Possessive wh-pronoun",
    "WRB": "Wh-adverb",
}
```

```python
# Store unique tokens

unique_tokens = {}

for token, tag in tagged_tokens:

    # Ignore punctuation
    if token in [".", ",", "!", "?", ";", ":"]:
        continue

    # Keep only the first occurrence
    if token not in unique_tokens:

        unique_tokens[token] = tag


# Create DataFrame

nltk_table = pd.DataFrame([
    {
        "Token": token,
        "POS Tag": tag,
        "Description": pos_descriptions.get(
            tag,
            "Description not available"
        )
    }
    for token, tag in unique_tokens.items()
])

nltk_table
```

# 5.4 — POS Tagging Using spaCy with Token Frequency

```python
!pip install spacy -q
```

```python
import spacy
import pandas as pd
from collections import Counter
```

```python
# Load the pretrained English NLP model

nlp = spacy.load("en_core_web_sm")

print("spaCy English model loaded successfully.")
```

```python
# Same input file used in Question 5.1

INPUT_FILE = "input_sub_word_data.txt"

with open(INPUT_FILE, "r", encoding="utf-8") as file:
    text = file.read()

print("Input data loaded successfully.")

print("\nFirst 1000 characters of input:\n")
print(text[:1000])
```

```python
# Process the complete input text

doc = nlp(text)

print("Text processed successfully.")

print("Total tokens including punctuation:", len(doc))
```

```python
# Count frequency of each token
# Convert tokens to lowercase for case-insensitive counting

token_frequency = Counter(
    token.text.lower()
    for token in doc
    if not token.is_space
)

print("Number of unique tokens:", len(token_frequency))
```

```python
# Store POS information for each unique token

unique_tokens = {}

for token in doc:

    # Ignore spaces
    if token.is_space:
        continue

    # Normalize token to lowercase
    token_text = token.text.lower()

    # Store information only once for each unique token
    if token_text not in unique_tokens:

        unique_tokens[token_text] = {
            "POS Tag": token.pos_,
            "Description": spacy.explain(token.pos_),
            "Frequency": token_frequency[token_text]
        }
```

```python
# Convert dictionary into a DataFrame

pos_frequency_table = pd.DataFrame([
    {
        "Unique Token": token,
        "POS Tag": information["POS Tag"],
        "Description": information["Description"],
        "Frequency": information["Frequency"]
    }

    for token, information in unique_tokens.items()
])

# Sort by frequency from highest to lowest
pos_frequency_table = pos_frequency_table.sort_values(
    by="Frequency",
    ascending=False
).reset_index(drop=True)

pos_frequency_table
```

```python
print("=" * 90)
print("POS TAGGING USING SPACY")
print("=" * 90)

print(
    f"{'TOKEN':<25}"
    f"{'POS TAG':<15}"
    f"{'DESCRIPTION':<40}"
    f"{'FREQUENCY'}"
)

print("-" * 90)

for _, row in pos_frequency_table.iterrows():

    print(
        f"{row['Unique Token']:<25}"
        f"{row['POS Tag']:<15}"
        f"{row['Description']:<40}"
        f"{row['Frequency']}"
    )

print("-" * 90)

print(
    "Total unique tokens:",
    len(pos_frequency_table)
)
```

## 5.7 Train a Word2Vec model using the provided corpus and calculate the cosine similarity between selected word pairs.

```python

from gensim.models import Word2Vec

# Read corpus
with open("corpus.txt", "r", encoding="utf-8") as file:
    sentences = [
        line.lower().split()
        for line in file
        if line.strip()
    ]

# Train Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    workers=1,
    seed=42,
    epochs=100
)

# Word pairs for similarity calculation
word_pairs = [
    ("machine", "learning"),
    ("natural", "language"),
    ("deep", "learning")
]

# Calculate cosine similarity
print("Cosine Similarity Results:\n")

for word1, word2 in word_pairs:
    similarity = model.wv.similarity(word1, word2)
    print(f"{word1} and {word2}: {similarity:.4f}")
```
