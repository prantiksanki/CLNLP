```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("employee_information.csv")
```

```python
df.head()
```

# Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.


```python
avg_salary_dept = df.groupby("Department")["Salary"].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(avg_salary_dept.index, avg_salary_dept.values, width=0.6)
ax.set_title("Average Salary by Department")
ax.set_xlabel("Department")
ax.set_ylabel("Average Salary")
ax.grid(axis="x", visible=False)
for b in bars:
    ax.annotate(f"{b.get_height():,.0f}", (b.get_x() + b.get_width() / 2, b.get_height()))
```

# Experiment 1.2: Display the number of employees working in each department.


```python
emp_count_dept = df["Department"].value_counts().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(emp_count_dept.index, emp_count_dept.values)
ax.set_title("Number of Employees per Department")
ax.set_xlabel("Department")
ax.set_ylabel("Employee Count")
for b in bars:
    ax.annotate(f"{int(b.get_height())}", (b.get_x() + b.get_width() / 2, b.get_height()))
```

# Experiment 1.3:Display the percentage of male and female employees using a pie chart


```python
gender_counts = df["Gender"].value_counts()

fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(
    gender_counts.values,
    labels=gender_counts.index
)
ax.set_title("Male vs Female Employees")
ax.axis("equal")
```

# Experiment 1.4: Draw a histogram showing salary distribution.

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(df["Salary"], bins=10, edgecolor="#fcfcfb")
ax.set_title("Salary Distribution")
ax.set_xlabel("Salary")
ax.set_ylabel("Number of Employees")
```

# Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df["Experience_Years"], df["Salary"], edgecolors="none")
ax.set_title("Experience vs Salary")
ax.set_xlabel("Experience (Years)")
ax.set_ylabel("Salary")
```

# Experiment 1.6: Display the top 10 employees with the highest salary.

```python
top10_salary = df.sort_values("Salary", ascending=False).head(10)
top10_salary[["Employee_ID", "Name", "Department", "Salary"]]
```

# Experiment 1.7: Find the highest salary in every department.

```python
max_salary_dept = df.groupby("Department")["Salary"].max().sort_values(ascending=False)
```

```python
max_salary_dept
```

# Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
overall_avg_salary = df["Salary"].mean()
above_avg = df[df["Salary"] > overall_avg_salary].sort_values("Salary", ascending=False)

print(f"Overall average salary: {overall_avg_salary:,.2f}")
print(f"Employees above average: {len(above_avg)}")
above_avg[["Employee_ID", "Name", "Department", "Salary"]]
```

# Experiment 1.9: Calculate the average years of experience in each department.

```python
avg_exp_dept = df.groupby("Department")["Experience_Years"].mean().sort_values(ascending=False)
avg_exp_dept
```

# Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(df["Age"], bins=10, edgecolor="#fcfcfb")
ax.set_title("Employee Age Distribution")
ax.set_xlabel("Age")
ax.set_ylabel("Number of Employees")
```
