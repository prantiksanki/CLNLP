# Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:

1. Convert text to lowercase. (Also display how many uppercase?)

```python
uppercase_count = sum(1 for ch in text if ch.isupper())
text = text.lower()

print("1. Number of uppercase letters:", uppercase_count)
```

2.	Remove punctuation.        (Also display how many different types of punctuations?)

```python
punctuations = [ch for ch in text if ch in string.punctuation]
different_punctuations = set(punctuations)

print("2. Number of different types of punctuations:",
      len(different_punctuations))
print("   Punctuations:", different_punctuations)

text = text.translate(str.maketrans("", "", string.punctuation))
```

3.	Remove numbers.            (Also display how many numbers?)

```python
numbers = re.findall(r'\d+', text)

print("3. Number of numbers:", len(numbers))

text = re.sub(r'\d+', '', text)
```

4.	Remove extra whitespace. (How much extra whitespace is present?)

```python
extra_whitespace = sum(
    len(match.group()) - 1
    for match in re.finditer(r'\s{2,}', text)
)

print("4. Amount of extra whitespace:", extra_whitespace)

text = re.sub(r'\s+', ' ', text).strip()
```

5.	Display the cleaned text.     


```python
print("\n5. Cleaned Text:")
print(text)
```

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize 
```

```python
# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# Read the text file
with open("2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
# Sentence Tokenization
sentences = sent_tokenize(text)
```

```python
# Word Tokenization
words = word_tokenize(text)
```

```python
print("SENTENCE TOKENIZATION:")
for i, sentence in enumerate(sentences, 1):
    print(i, ".", sentence)
```

```python
print("\nWORD TOKENIZATION:")
print(words)
```

# Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
# Read the text file
with open("2.3_clean_data.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
# Tokenize the text into words
tokens = re.findall(r'\b\w+\b', text.lower())
```

```python
# Define common stop words
stop_words = {
    "is", "the", "a", "an", "in", "of", "and", "to",
    "for", "on", "with", "from", "by", "this", "that",
    "it", "as", "such", "are", "was", "were", "be",
    "can", "one", "some", "these", "those", "or"
}
```

```python
# Find stop words present in the text
found_stop_words = [word for word in tokens if word in stop_words]

# Remove stop words
filtered_tokens = [word for word in tokens if word not in stop_words]
```

```python
# Display original tokens
print("ORIGINAL TOKENS:")
print(tokens)
```

```python
# Display stop words separately
print("STOP WORDS:")
print(found_stop_words)
```

```python
# Display filtered tokens
print("FILTERED TOKENS:")
print(filtered_tokens)
```
