# Experiment 3.1:  Write a Python program to perform  Write a Python program to perform stemming using the Porter Stemmer. Apply stemming to the following words. Display the original words and their stems.
playing, played, plays, studies, studying, connected, connection, computers



```python
import nltk
from nltk.stem import PorterStemmer
```

```python
# Create stemmer
stemmer = PorterStemmer()

words = [
    "playing", "played", "plays", "studies",
    "studying", "connected", "connection", "computers"
]
```

```python
print("Original Word -> Stem")
print("----------------------")
for word in words:
    stem = stemmer.stem(word)
    print(word, "->", stem)
```

# Experiment 3.2:  Write a Python program to perform lemmatization on a given input words:

cats,	dogs,   running,  runs,	ran, studies, studying,	better, children, mice,	went, ate,
leaves,	caring,


```python
import nltk

nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = [
    "cats", "dogs", "running", "runs", "ran",
    "studies", "studying", "better", "children",
    "mice", "went", "ate", "leaves", "caring"
]
```

```python
print("Original Word -> Lemma")
print("----------------------")

for word in words:

    # Use noun for nouns
    if word in ["cats", "dogs", "children", "mice", "leaves"]:
        lemma = lemmatizer.lemmatize(word, pos="n")

    # Use adjective for better
    elif word == "better":
        lemma = lemmatizer.lemmatize(word, pos="a")

    # Use verb for remaining words
    else:
        lemma = lemmatizer.lemmatize(word, pos="v")

    print(word, "->", lemma)
```

# Experiment 3.3: Write a Python program using Regular Expressions (Regex) to extract and process the following information from a text:
# Email addresses, URLs, Mobile numbers, Hashtags, Mentions


```python
import re

text = """Welcome to the Natural Language Processing workshop!
For more information, visit https://www.nlpworkshop.com or www.python.org.
You can contact the coordinator at nlpworkshop@gmail.com or support@python.org.
For registration, call +91-9876543210 or 9123456789.
Follow us on social media @NLPWorkshop and @PythonLearner.
Share your experience using #NLP, #Python, and #MachineLearning.
You can also visit https://github.com/NLPWorkshop for the latest updates."""

# Find email addresses
emails = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text)

# Find URLs
urls = re.findall(r'https?://\S+|www\.\S+', text)

# Find mobile numbers
mobile_numbers = re.findall(r'(?:\+91[- ]?)?[6-9]\d{9}', text)

# Find hashtags
hashtags = re.findall(r'#\w+', text)

# Find mentions
mentions = re.findall(r'@\w+', text)


print("Email Addresses:")
print(emails)

print("\nURLs:")
print(urls)

print("\nMobile Numbers:")
print(mobile_numbers)

print("\nHashtags:")
print(hashtags)

print("\nMentions:")
print(mentions)
```
